```python
from transformers import PegasusForConditionalGeneration, PegasusTokenizer
import torch
```

```python
model_name = "tuner007/pegasus_paraphrase"
tokenizer = PegasusTokenizer.from_pretrained(model_name)
model = PegasusForConditionalGeneration.from_pretrained(model_name)
```

```python
def simplify_text(text):
    batch = tokenizer([text], truncation=True, padding="longest", return_tensors="pt")
    translated = model.generate(**batch, max_length=60, num_beams=5, early_stopping=True)
    simplified = tokenizer.batch_decode(translated, skip_special_tokens=True)[0]
    return simplified
```

```python
simplify_text("Photosynthesis is the process by which green plants convert sunlight into chemical energy.")
```

```python
qa_model = pipeline(
    "question-answering",
    model="distilbert-base-cased-distilled-squad",
    tokenizer="distilbert-base-cased-distilled-squad"
)
```

```python
def answer_question(question, context):
    result = qa_model(question=question, context=context)
    return result["answer"]
```

```python
context = "The Sun is the main source of energy for life on Earth."
question = "What is the main source of energy for Earth?"
answer_question(question, context)
```
